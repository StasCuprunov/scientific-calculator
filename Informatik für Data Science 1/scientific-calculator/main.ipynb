```python
import tkinter as tk
from constants import *

_button_size = 5
_padding = 10

root = tk.Tk()
root.title("Taschenrechner")

def button_zero():
    text_input.set(text_input.get() + CHARACTER_ZERO)

def button_one():
    text_input.set(text_input.get() + CHARACTER_ONE)

def button_two():
    text_input.set(text_input.get() + CHARACTER_TWO)

def button_three():
    text_input.set(text_input.get() + CHARACTER_THREE)

def button_four():
    text_input.set(text_input.get() + CHARACTER_FOUR)

def button_five():
    text_input.set(text_input.get() + CHARACTER_FIVE)

def button_six():
    text_input.set(text_input.get() + CHARACTER_SIX)

def button_seven():
    text_input.set(text_input.get() + CHARACTER_SEVEN)

def button_eight():
    text_input.set(text_input.get() + CHARACTER_EIGHT)

def button_nine():
    text_input.set(text_input.get() + CHARACTER_NINE)

def button_delete():
    length = len(text_input.get())
    text_input.set(text_input.get()[:length - 1])

def button_minus():
    text_input.set(text_input.get() + CHARACTER_MINUS)

def button_decimal_point():
    text_input.set(text_input.get() + CHARACTER_DECIMAL_POINT)

def button_calculate():
    print(f"Calculate {text_input.get()}")

text_input = tk.StringVar()

label = tk.Label(root, 
                 textvariable=text_input,        
                 height=6,              
                 width=40,                             
                 font=("Arial", 10, "bold"),
                 fg="red",                        
                 relief=tk.RAISED,     
                 underline=None,           
                 wraplength=100       
                ).grid(row=0, column=1, rowspan=1, columnspan=5, padx=_padding , pady=_padding)

button_delete = tk.Button(root, text=CHARACTER_DELETE, command=button_delete, width=_button_size, height=_button_size)
button_delete.grid(row=1, column=4, rowspan=1, padx=_padding, pady=_padding)



button_seven = tk.Button(root, text=CHARACTER_SEVEN, command=button_seven, width=_button_size, height=_button_size)
button_seven.grid(row=2, column=1, rowspan=1, padx=_padding, pady=_padding)

button_eight = tk.Button(root, text=CHARACTER_EIGHT, command=button_eight, width=_button_size, height=_button_size)
button_eight.grid(row=2, column=2, rowspan=1, padx=_padding, pady=_padding)

button_nine = tk.Button(root, text=CHARACTER_NINE, command=button_nine, width=_button_size, height=_button_size)
button_nine.grid(row=2, column=3, rowspan=1, padx=_padding, pady=_padding)




button_four = tk.Button(root, text=CHARACTER_FOUR, command=button_four,width=_button_size, height=_button_size)
button_four.grid(row=3, column=1, rowspan=1, padx=_padding, pady=_padding)

button_five = tk.Button(root, text=CHARACTER_FIVE, command=button_five, width=_button_size, height=_button_size)
button_five.grid(row=3, column=2, rowspan=1, padx=_padding, pady=_padding)

button_six = tk.Button(root, text=CHARACTER_SIX, command=button_six, width=_button_size, height=_button_size)
button_six.grid(row=3, column=3, rowspan=1, padx=_padding, pady=_padding)




button_one = tk.Button(root, text=CHARACTER_ONE, command=button_one, width=_button_size, height=_button_size)
button_one.grid(row=4, column=1, rowspan=1, padx=_padding, pady=_padding)

button_two = tk.Button(root, text=CHARACTER_TWO, command=button_two, width=_button_size, height=_button_size)
button_two.grid(row=4, column=2, rowspan=1, padx=_padding, pady=_padding)

button_three = tk.Button(root, text=CHARACTER_THREE, command=button_three, width=_button_size, height=_button_size)
button_three.grid(row=4, column=3, rowspan=1, padx=_padding, pady=_padding)



button_minus = tk.Button(root, text=CHARACTER_MINUS, command=button_minus, width=_button_size, height=_button_size)
button_minus.grid(row=5, column=1, rowspan=1, padx=_padding, pady=_padding)

button_zero = tk.Button(root, text=CHARACTER_ZERO, command=button_zero, width=_button_size, height=_button_size)
button_zero.grid(row=5, column=2, rowspan=1, padx=_padding, pady=_padding)

button_decimal_point = tk.Button(root, text=CHARACTER_DECIMAL_POINT, command=button_decimal_point, width=_button_size, height=_button_size)
button_decimal_point.grid(row=5, column=3, rowspan=1, padx=_padding, pady=_padding)

button_calculate = tk.Button(root, text=CHARACTER_CALCULATE, command=button_calculate, width=_button_size, height=_button_size)
button_calculate.grid(row=5, column=4, rowspan=1, padx=_padding, pady=_padding)

root.mainloop()
```
