```python
import tkinter as tk

root = tk.Tk()
root.title("Taschenrechner")

def button_zero():
    text_input.set(text_input.get() + "0")

def button_one():
    text_input.set(text_input.get() + "1")

def button_two():
    text_input.set(text_input.get() + "2")

def button_three():
    text_input.set(text_input.get() + "3")

def button_four():
    text_input.set(text_input.get() + "4")

def button_five():
    text_input.set(text_input.get() + "5")

def button_six():
    text_input.set(text_input.get() + "6")

def button_seven():
    text_input.set(text_input.get() + "7")

def button_eight():
    text_input.set(text_input.get() + "8")

def button_nine():
    text_input.set(text_input.get() + "9")

def button_delete():
    length = len(text_input.get())
    text_input.set(text_input.get()[:length - 1])

def button_minus():
    text_input.set(text_input.get() + "-")

def button_decimal_point():
    text_input.set(text_input.get() + ".")

def button_calculate():
    print(f"Calculate {text_input.get()}")

text_input = tk.StringVar()

label = tk.Label(root, 
                 textvariable=text_input, 
                 anchor=tk.CENTER,       
                 bg="lightblue",      
                 height=3,              
                 width=10,              
                 bd=3,                  
                 font=("Arial", 10, "bold"), 
                 cursor="hand2",   
                 fg="red",                        
                 justify=tk.CENTER,    
                 relief=tk.RAISED,     
                 underline=0,           
                 wraplength=100        
                ).grid(row=0, column=0, rowspan=1, padx=10, pady=10)


button_delete = tk.Button(root, text="<-", command=button_delete)
button_delete.grid(row=1, column=2, rowspan=1, padx=10, pady=10)



button_seven = tk.Button(root, text="7", command=button_seven)
button_seven.grid(row=2, column=0, rowspan=1, padx=10, pady=10)

button_eight = tk.Button(root, text="8", command=button_eight)
button_eight.grid(row=2, column=1, rowspan=1, padx=10, pady=10)

button_nine = tk.Button(root, text="9", command=button_nine)
button_nine.grid(row=2, column=2, rowspan=1, padx=10, pady=10)




button_four = tk.Button(root, text="4", command=button_four)
button_four.grid(row=3, column=0, rowspan=1, padx=10, pady=10)

button_five = tk.Button(root, text="5", command=button_five)
button_five.grid(row=3, column=1, rowspan=1, padx=10, pady=10)

button_six = tk.Button(root, text="6", command=button_six)
button_six.grid(row=3, column=2, rowspan=1, padx=10, pady=10)




button_one = tk.Button(root, text="1", command=button_one)
button_one.grid(row=4, column=0, rowspan=1, padx=10, pady=10)

button_two = tk.Button(root, text="2", command=button_two)
button_two.grid(row=4, column=1, rowspan=1, padx=10, pady=10)

button_three = tk.Button(root, text="3", command=button_three)
button_three.grid(row=4, column=2, rowspan=1, padx=10, pady=10)



button_minus = tk.Button(root, text="-", command=button_minus)
button_minus.grid(row=5, column=0, rowspan=1, padx=10, pady=10)

button_zero = tk.Button(root, text="0", command=button_zero)
button_zero.grid(row=5, column=1, rowspan=1, padx=10, pady=10)

button_decimal_point = tk.Button(root, text=".", command=button_decimal_point)
button_decimal_point.grid(row=5, column=2, rowspan=1, padx=10, pady=10)

button_calculate = tk.Button(root, text="=", command=button_calculate)
button_calculate.grid(row=5, column=3, rowspan=1, padx=10, pady=10)

root.mainloop()
```
