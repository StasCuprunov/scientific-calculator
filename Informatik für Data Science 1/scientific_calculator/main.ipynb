```python
import tkinter as tk

import sys
sys.path.append("constants")
from character_constants import *
from text_constants import *
from style_constants import *

sys.path.append("calculation")
from calculation import *
from load_constants import *

def center_window(width, height):
    screen_width = root.winfo_screenwidth()
    screen_height = root.winfo_screenheight()

    x = (screen_width/2) - (width/2)
    y = (screen_height/2) - (height/2)
    root.geometry('%dx%d+%d+%d' % (width, height, x, y))

root = tk.Tk()
center_window(WINDOW_WIDTH, WINDOW_WIDTH * 2)
root.title("Taschenrechner")

text_input = tk.StringVar()
store_variable = 0

constants_for_calculation = get_constants_for_calculation()


def button_add_zero():
    text_input.set(text_input.get() + CHARACTER_ZERO)

def button_add_one():
    text_input.set(text_input.get() + CHARACTER_ONE)

def button_add_two():
    text_input.set(text_input.get() + CHARACTER_TWO)

def button_add_three():
    text_input.set(text_input.get() + CHARACTER_THREE)

def button_add_four():
    text_input.set(text_input.get() + CHARACTER_FOUR)

def button_add_five():
    text_input.set(text_input.get() + CHARACTER_FIVE)

def button_add_six():
    text_input.set(text_input.get() + CHARACTER_SIX)

def button_add_seven():
    text_input.set(text_input.get() + CHARACTER_SEVEN)

def button_add_eight():
    text_input.set(text_input.get() + CHARACTER_EIGHT)

def button_add_nine():
    text_input.set(text_input.get() + CHARACTER_NINE)

def button_clear():
    text_input.set("")

def button_delete():
    if (text_input.get() == TEXT_INVALID_SYNTAX):
        text_input.set("")
    length = len(text_input.get())
    text_input.set(text_input.get()[:length - 1])

def button_store_variable():
    global store_variable
    store_variable = text_input.get()

def button_use_variable():
    text_input.set(text_input.get() + str(store_variable))

def button_add_minus():
    text_input.set(text_input.get() + CHARACTER_MINUS)

def button_add_plus():
    text_input.set(text_input.get() + CHARACTER_PLUS)

def button_add_multiply():
    text_input.set(text_input.get() + CHARACTER_MULTIPLY)

def button_add_division():
    text_input.set(text_input.get() + CHARACTER_DIVISION)

def button_add_decimal_point():
    text_input.set(text_input.get() + CHARACTER_DECIMAL_POINT)

def menu_add_constant(key):
    text_input.set(text_input.get() + str(constants_for_calculation[key]))

def on_enter_constants(event, item_widget):
    item_widget.config(relief="sunken", borderwidth=1)

def on_leave_constants(event, item_widget):
    item_widget.config(relief="raised", borderwidth=1)

def button_calculate():
    text_input.set(calculate(text_input.get()))



label = tk.Label(root, 
                 textvariable=text_input,        
                 height=6,              
                 width=40,                             
                 font=("Arial", 10, "bold"),                        
                 relief=tk.RAISED,     
                 underline=None,           
                 wraplength=100       
                ).grid(row=0, column=1, rowspan=1, columnspan=5, padx=PADDING , pady=PADDING)

button_constants = tk.Menubutton(root, text="Konstanten", width=BUTTON_WIDTH*2, height=BUTTON_HEIGHT, relief=tk.RAISED)
button_constants.bind("<Enter>", lambda e: on_enter_constants(e, button_constants))
button_constants.bind("<Leave>", lambda e: on_leave_constants(e, button_constants))
button_constants.grid(row=1, column=1, rowspan=1, padx=PADDING, pady=PADDING)
button_constants.menu = tk.Menu(button_constants, tearoff = 0)
button_constants["menu"] = button_constants.menu

for key in constants_for_calculation.keys():
    button_constants.menu.add_cascade(label=key, command=lambda element=key: menu_add_constant(element))

button_clear = tk.Button(root, text=CHARACTER_CLEAR, command=button_clear, width=BUTTON_WIDTH, height=BUTTON_HEIGHT)
button_clear.grid(row=1, column=3, rowspan=1, padx=PADDING, pady=PADDING)

button_delete = tk.Button(root, text=CHARACTER_DELETE, command=button_delete, width=BUTTON_WIDTH, height=BUTTON_HEIGHT)
button_delete.grid(row=1, column=4, rowspan=1, padx=PADDING, pady=PADDING)



button_store_variable = tk.Button(root, text="Speichern", command=button_store_variable, width=BUTTON_WIDTH*2, height=BUTTON_HEIGHT)
button_store_variable.grid(row=2, column=1, rowspan=1, padx=PADDING, pady=PADDING)

button_use_variable = tk.Button(root, text="Benutzen", command=button_use_variable, width=BUTTON_WIDTH*2, height=BUTTON_HEIGHT)
button_use_variable.grid(row=2, column=2, rowspan=1, padx=PADDING, pady=PADDING)



button_seven = tk.Button(root, text=CHARACTER_SEVEN, command=button_add_seven, width=BUTTON_WIDTH,height=BUTTON_HEIGHT)
button_seven.grid(row=3, column=1, rowspan=1, padx=PADDING, pady=PADDING)

button_eight = tk.Button(root, text=CHARACTER_EIGHT, command=button_add_eight, width=BUTTON_WIDTH, height=BUTTON_HEIGHT)
button_eight.grid(row=3, column=2, rowspan=1, padx=PADDING, pady=PADDING)

button_nine = tk.Button(root, text=CHARACTER_NINE, command=button_add_nine, width=BUTTON_WIDTH, height=BUTTON_HEIGHT)
button_nine.grid(row=3, column=3, rowspan=1, padx=PADDING, pady=PADDING)

button_division = tk.Button(root, text=CHARACTER_DIVISION, command=button_add_division, width=BUTTON_WIDTH, height=BUTTON_HEIGHT)
button_division.grid(row=3, column=4, rowspan=1, padx=PADDING, pady=PADDING)


button_four = tk.Button(root, text=CHARACTER_FOUR, command=button_add_four,width=BUTTON_WIDTH, height=BUTTON_HEIGHT)
button_four.grid(row=4, column=1, rowspan=1, padx=PADDING, pady=PADDING)

button_five = tk.Button(root, text=CHARACTER_FIVE, command=button_add_five, width=BUTTON_WIDTH, height=BUTTON_HEIGHT)
button_five.grid(row=4, column=2, rowspan=1, padx=PADDING, pady=PADDING)

button_six = tk.Button(root, text=CHARACTER_SIX, command=button_add_six, width=BUTTON_WIDTH, height=BUTTON_HEIGHT)
button_six.grid(row=4, column=3, rowspan=1, padx=PADDING, pady=PADDING)

button_multiply = tk.Button(root, text=CHARACTER_MULTIPLY, command=button_add_multiply, width=BUTTON_WIDTH, height=BUTTON_HEIGHT)
button_multiply.grid(row=4, column=4, rowspan=1, padx=PADDING, pady=PADDING)



button_one = tk.Button(root, text=CHARACTER_ONE, command=button_add_one, width=BUTTON_WIDTH, height=BUTTON_HEIGHT)
button_one.grid(row=5, column=1, rowspan=1, padx=PADDING, pady=PADDING)

button_two = tk.Button(root, text=CHARACTER_TWO, command=button_add_two, width=BUTTON_WIDTH, height=BUTTON_HEIGHT)
button_two.grid(row=5, column=2, rowspan=1, padx=PADDING, pady=PADDING)

button_three = tk.Button(root, text=CHARACTER_THREE, command=button_add_three, width=BUTTON_WIDTH, height=BUTTON_HEIGHT)
button_three.grid(row=5, column=3, rowspan=1, padx=PADDING, pady=PADDING)

button_plus = tk.Button(root, text=CHARACTER_PLUS, command=button_add_plus, width=BUTTON_WIDTH, height=BUTTON_HEIGHT)
button_plus.grid(row=5, column=4, rowspan=1, padx=PADDING, pady=PADDING)


button_minus = tk.Button(root, text=CHARACTER_MINUS, command=button_add_minus, width=BUTTON_WIDTH, height=BUTTON_HEIGHT)
button_minus.grid(row=6, column=1, rowspan=1, padx=PADDING, pady=PADDING)

button_zero = tk.Button(root, text=CHARACTER_ZERO, command=button_add_zero, width=BUTTON_WIDTH, height=BUTTON_HEIGHT)
button_zero.grid(row=6, column=2, rowspan=1, padx=PADDING, pady=PADDING)

button_decimal_point = tk.Button(root, text=CHARACTER_DECIMAL_POINT, command=button_add_decimal_point, width=BUTTON_WIDTH, height=BUTTON_HEIGHT)
button_decimal_point.grid(row=6, column=3, rowspan=1, padx=PADDING, pady=PADDING)

button_calculate = tk.Button(root, text=CHARACTER_CALCULATE, command=button_calculate, width=BUTTON_WIDTH, height=BUTTON_HEIGHT)
button_calculate.grid(row=6, column=4, rowspan=1, padx=PADDING, pady=PADDING)

root.mainloop()
```
