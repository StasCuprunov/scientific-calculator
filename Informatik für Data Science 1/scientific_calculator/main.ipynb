```python
import tkinter as tk

import sys
sys.path.append("constants")
from character_constants import *
from text_constants import *
from style_constants import *

sys.path.append("calculation")
from calculation import *

def center_window(width, height):
    screen_width = root.winfo_screenwidth()
    screen_height = root.winfo_screenheight()

    x = (screen_width/2) - (width/2)
    y = (screen_height/2) - (height/2)
    root.geometry('%dx%d+%d+%d' % (width, height, x, y))

root = tk.Tk()
center_window(WINDOW_WIDTH, WINDOW_WIDTH * 2)
root.title("Taschenrechner")

text_input = tk.StringVar()


def button_add_zero():
    text_input.set(text_input.get() + CHARACTER_ZERO)

def button_add_one():
    text_input.set(text_input.get() + CHARACTER_ONE)

def button_add_two():
    text_input.set(text_input.get() + CHARACTER_TWO)

def button_add_three():
    text_input.set(text_input.get() + CHARACTER_THREE)

def button_add_four():
    text_input.set(text_input.get() + CHARACTER_FOUR)

def button_add_five():
    text_input.set(text_input.get() + CHARACTER_FIVE)

def button_add_six():
    text_input.set(text_input.get() + CHARACTER_SIX)

def button_add_seven():
    text_input.set(text_input.get() + CHARACTER_SEVEN)

def button_add_eight():
    text_input.set(text_input.get() + CHARACTER_EIGHT)

def button_add_nine():
    text_input.set(text_input.get() + CHARACTER_NINE)

def button_delete():
    if (text_input.get() == TEXT_INVALID_SYNTAX):
        text_input.set("")
    length = len(text_input.get())
    text_input.set(text_input.get()[:length - 1])

def button_add_minus():
    text_input.set(text_input.get() + CHARACTER_MINUS)

def button_add_plus():
    text_input.set(text_input.get() + CHARACTER_PLUS)

def button_add_multiply():
    text_input.set(text_input.get() + CHARACTER_MULTIPLY)

def button_add_division():
    text_input.set(text_input.get() + CHARACTER_DIVISION)

def button_add_decimal_point():
    text_input.set(text_input.get() + CHARACTER_DECIMAL_POINT)

def button_calculate():
    text_input.set(calculate(text_input.get()))

label = tk.Label(root, 
                 textvariable=text_input,        
                 height=6,              
                 width=40,                             
                 font=("Arial", 10, "bold"),                        
                 relief=tk.RAISED,     
                 underline=None,           
                 wraplength=100       
                ).grid(row=0, column=1, rowspan=1, columnspan=5, padx=PADDING , pady=PADDING)


button_delete = tk.Button(root, text=CHARACTER_DELETE, command=button_delete, width=BUTTON_SIZE, height=BUTTON_SIZE)
button_delete.grid(row=1, column=4, rowspan=1, padx=PADDING, pady=PADDING)


button_seven = tk.Button(root, text=CHARACTER_SEVEN, command=button_add_seven, width=BUTTON_SIZE, height=BUTTON_SIZE)
button_seven.grid(row=2, column=1, rowspan=1, padx=PADDING, pady=PADDING)

button_eight = tk.Button(root, text=CHARACTER_EIGHT, command=button_add_eight, width=BUTTON_SIZE, height=BUTTON_SIZE)
button_eight.grid(row=2, column=2, rowspan=1, padx=PADDING, pady=PADDING)

button_nine = tk.Button(root, text=CHARACTER_NINE, command=button_add_nine, width=BUTTON_SIZE, height=BUTTON_SIZE)
button_nine.grid(row=2, column=3, rowspan=1, padx=PADDING, pady=PADDING)

button_division = tk.Button(root, text=CHARACTER_DIVISION, command=button_add_division, width=BUTTON_SIZE, height=BUTTON_SIZE)
button_division.grid(row=2, column=4, rowspan=1, padx=PADDING, pady=PADDING)


button_four = tk.Button(root, text=CHARACTER_FOUR, command=button_add_four,width=BUTTON_SIZE, height=BUTTON_SIZE)
button_four.grid(row=3, column=1, rowspan=1, padx=PADDING, pady=PADDING)

button_five = tk.Button(root, text=CHARACTER_FIVE, command=button_add_five, width=BUTTON_SIZE, height=BUTTON_SIZE)
button_five.grid(row=3, column=2, rowspan=1, padx=PADDING, pady=PADDING)

button_six = tk.Button(root, text=CHARACTER_SIX, command=button_add_six, width=BUTTON_SIZE, height=BUTTON_SIZE)
button_six.grid(row=3, column=3, rowspan=1, padx=PADDING, pady=PADDING)

button_multiply = tk.Button(root, text=CHARACTER_MULTIPLY, command=button_add_multiply, width=BUTTON_SIZE, height=BUTTON_SIZE)
button_multiply.grid(row=3, column=4, rowspan=1, padx=PADDING, pady=PADDING)



button_one = tk.Button(root, text=CHARACTER_ONE, command=button_add_one, width=BUTTON_SIZE, height=BUTTON_SIZE)
button_one.grid(row=4, column=1, rowspan=1, padx=PADDING, pady=PADDING)

button_two = tk.Button(root, text=CHARACTER_TWO, command=button_add_two, width=BUTTON_SIZE, height=BUTTON_SIZE)
button_two.grid(row=4, column=2, rowspan=1, padx=PADDING, pady=PADDING)

button_three = tk.Button(root, text=CHARACTER_THREE, command=button_add_three, width=BUTTON_SIZE, height=BUTTON_SIZE)
button_three.grid(row=4, column=3, rowspan=1, padx=PADDING, pady=PADDING)

button_plus = tk.Button(root, text=CHARACTER_PLUS, command=button_add_plus, width=BUTTON_SIZE, height=BUTTON_SIZE)
button_plus.grid(row=4, column=4, rowspan=1, padx=PADDING, pady=PADDING)


button_minus = tk.Button(root, text=CHARACTER_MINUS, command=button_add_minus, width=BUTTON_SIZE, height=BUTTON_SIZE)
button_minus.grid(row=5, column=1, rowspan=1, padx=PADDING, pady=PADDING)

button_zero = tk.Button(root, text=CHARACTER_ZERO, command=button_add_zero, width=BUTTON_SIZE, height=BUTTON_SIZE)
button_zero.grid(row=5, column=2, rowspan=1, padx=PADDING, pady=PADDING)

button_decimal_point = tk.Button(root, text=CHARACTER_DECIMAL_POINT, command=button_add_decimal_point, width=BUTTON_SIZE, height=BUTTON_SIZE)
button_decimal_point.grid(row=5, column=3, rowspan=1, padx=PADDING, pady=PADDING)

button_calculate = tk.Button(root, text=CHARACTER_CALCULATE, command=button_calculate, width=BUTTON_SIZE, height=BUTTON_SIZE)
button_calculate.grid(row=5, column=4, rowspan=1, padx=PADDING, pady=PADDING)

root.mainloop()
```
