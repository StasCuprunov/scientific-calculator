```python
import unittest

import sys
sys.path.append("../calculation")
from calculation import *

class UnitTests(unittest.TestCase):
    def test_addition_one(self):
        result = calculate("2+3")
        self.assertEqual(result, str(5.0))

    def test_addition_two(self):
        result = calculate("-3+4")
        self.assertEqual(result, str(1.0))

    def test_substraction_one(self):
        result = calculate("-2.6-7.5")
        self.assertEqual(result, str(-10.1))

    def test_substraction_two(self):
        result = calculate("10-7.91")
        self.assertEqual(result, str(2.09))

unittest.main(argv=[''], verbosity=2, exit=False)
```
