```python
import unittest

import sys
sys.path.append("../calculation")
from calculation import *

class UnitTests(unittest.TestCase):
    def test_addition_one(self):
        result = calculate("2+3")
        self.assertEqual(result, 5.0)

    def test_addition_two(self):
        result = calculate("-3+4")
        self.assertEqual(result, 1.0)

    def test_substraction_one(self):
        result = calculate("-2.6-7.5")
        self.assertEqual(result, -10.1)

    def test_substraction_two(self):
        result = calculate("10-7.91")
        self.assertEqual(result, 2.09)

    def test_multiply_one(self):
        result = calculate("5*6")
        self.assertEqual(result, 30.0)

    def test_multiply_two(self):
        result = calculate("-2*4.5")
        self.assertEqual(result, -9.0)

    def test_division_one(self):
        result = calculate("-4/2")
        self.assertEqual(result, -2.0)

    def test_division_two(self):
        result = calculate("10/4")
        self.assertEqual(result, 2.5)

    def test_division_three(self):
        result = calculate("42.0/42")
        self.assertEqual(result, 1.0)

    def test_arithmetic_operation_one(self):
        result = calculate("-3*5+0.5-6/2*5")
        self.assertEqual(result, -29.5)

    def test_arithmetic_operation_two(self):
        result = calculate("10*0.5*3-42.1+9-4*4/4/2")
        self.assertEqual(result, -20.1)

unittest.main(argv=[''], verbosity=2, exit=False)
```
